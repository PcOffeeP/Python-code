## 原题题意整理

给定 n 行 m 列的网格，0 表示可走的位置，1 表示障碍。只能上下左右移动，不能斜走或越出网格。

从左上角 (0,0) 走到右下角 (n-1,m-1)，输出完整路径。起点和终点可走，题目保证存在且只有一条可行路径。

**输入：**第一行是 n、m；随后 n 行，每行 m 个 0 或 1。

**输出：**按行走顺序每行一个坐标 (行,列)，从 0 开始编号，包含起点和终点，格式如 (0,0)。不是只输出步数。

## 原题题意整理

依次判断输入的密码。满足以下全部条件才合格：

1. 长度至少为 8。
2. 大写字母、小写字母、数字、其他特殊字符这四类中，至少出现三类。
3. 不存在两段互不重叠、长度至少为 3、内容完全相同的连续子串。重叠的相同子串不单独构成违规。

**输入：**若干行密码，读到 EOF（文件结束）；没有单独的组数行。原站范围为 1～10 组，每个密码长 1～100，只含 ASCII 33～126 的可见字符，不含空格或换行。

**输出：**每个密码对应一行，合格输出 OK，否则输出 NG。

## 原站样例（节选）

输入：

```text
021Abc9000
021Abc9Abc1
021ABC9000
021$bc9000
021Abc1111
```

输出：

```text
OK
NG
NG
OK
OK
```

## 原题题意整理

每条记录由一个索引和一个数值组成。将索引相同的记录汇总成一条，数值等于这些记录数值的总和；最后按索引升序列出结果。

**输入：**第一行是记录数 n；随后 n 行，每行两个整数 index、value。

**输出：**每行输出一个索引及对应的汇总值，中间用空格分隔。

## 原站样例（节选）

输入：

```text
4
0 1
0 2
1 2
3 4
```

输出：

```text
0 3
1 2
3 4
```

## 初步提醒

1. 你要保存的是“索引 → 累计数值”，不是输入行号与数值的对应关系。
2. 同一个索引再次出现时要累加；第一次出现时想清楚从什么初值开始。
3. 最后按索引排序，不是按累计数值排序，也不是按第一次出现的顺序输出。

## 练习区

先运行样例数据单元，再在空白单元编写。可以直接使用循环，不强制封装函数。

下方只提供输入数据，没有判题器；`expected_output` / `expected_result` 只是供你比对的预期结果。

````python
# 本地样例；可用 sample_input.splitlines() 练习解析。
# 正式提交时需改为标准输入，不能把这个样例写死。
sample_input = "4\n0 1\n0 2\n1 2\n3 4\n"
expected_output = "0 3\n1 2\n3 4\n"
````

````python
# 在这里写你的代码。
n = int(input())
sample = {}

for _ in range(n):
    a, b = map(int, input().split())
    sample[a] = sample.get(a, 0) + b

for i in sorted(sample.keys()):
    print(i, sample[i])
````

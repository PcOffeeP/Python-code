## 原题题意整理

给出 n 个整数。去掉重复项，使每个数只出现一次，再按数值升序输出。

**输入：**第一行是 n，接下来 n 行各有一个整数。

**输出：**每行一个整数，从小到大列出所有不同的数；不要输出列表括号或说明文字。

## 原站样例（节选）

输入：

```text
3
2
2
1
```

输出：

```text
1
2
```

````python
# 在这里写你的代码。
total_number = int(input())
number_set = set()

for i in range(total_number):
    # 添加新元素
    number_set.add(int(input()))

# 整理
result = sorted(number_set)

for x in result:
    print(x)
````

## 原题题意整理

从 (0,0) 出发，按顺序执行分号分隔的移动指令。A、D、W、S 分别表示左、右、上、下：左右改变 x，上下改变 y。

合法指令由方向字母、1～2 位十进制数字和结尾分号构成；移动距离是 1～99。小于 10 的距离允许一个前导零，例如 S01。非法片段直接忽略，不影响后续处理。

**输入：**一行由大写字母、数字和分号组成的指令串，至少含一个分号，且以分号结尾。

**输出：**最终坐标，格式为 x,y，不带括号。

## 原站样例（节选）

```text

```

输出：

```text

```
